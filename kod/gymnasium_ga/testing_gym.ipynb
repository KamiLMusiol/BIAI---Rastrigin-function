```python
import gymnasium as gym

# mozna zmienic None na 'human' na widowsie wtedy powinien byc teoretycznie int
env = gym.make("Blackjack-v1", render_mode=None)

while True:
    observation, info = env.reset()
    terminated = False
    truncated = False



    while not (terminated or truncated):
        player_sum, dealer_card, usable_ace = observation
        print(f"Twoje punkty: {player_sum} | Karta krupiera: {dealer_card} | As: {usable_ace}")

        move = input("1 - Dobierz (Hit), 0 - Czekaj (Stand): ")
        action = int(move)

        observation, reward, terminated, truncated, info = env.step(action)

        if terminated or truncated:
            print(f"WYNIK: {reward}")

    if input("Jeszcze raz? (t/n): ").lower() != 't':
        break
env.close()
```
